```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import Tool
```

```python
load_dotenv(dotenv_path=".env")

api_key = os.getenv("OPENAI_API_KEY")

print(type(api_key))
print(api_key is not None)
```

```python
llm=ChatOpenAI(
    model="gpt-6-luna",
    api_key=api_key
)
```

```python
#CREATE CHAT PROMPT TEMPLATE

prompt=ChatPromptTemplate.from_template("Answer clearly: {question}")

#Create the LCEL chain (LangChain Expression Language).
qa_chain=prompt | llm

#wrap LCEL chain as a tool
qa_tool=Tool(
    name="Simple QA",
    func=lambda question:qa_chain.invoke({"question":question}).content,
        #Lambda is simply a concise way to adapt the tool's string input to the dictionary input expected by the LCEL chain and to return only the generated text.
    description="A basic LLM chain that anwsers claear questions"
)
```

```python
#Use the tool

query="What is LangGraph in LangChain?"

answer=qa_tool.invoke(query)

#output
print("\n User Question:",query)
print("\n Tool Answer:",answer)
```
